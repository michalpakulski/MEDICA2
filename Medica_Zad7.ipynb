```python
import sqlite3
from IPython.display import display, HTML
import ipywidgets as widgets

# Utworzenie połączenia z wbudowaną bazą SQLite
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Tworzenie tabeli pacjentów
cursor.execute('''
CREATE TABLE patients (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    surname TEXT,
    pesel TEXT UNIQUE,
    email TEXT,
    password TEXT
)
''')
conn.commit()
```

```python
def add_patient_form():
    name = widgets.Text(description='Imię:')
    surname = widgets.Text(description='Nazwisko:')
    pesel = widgets.Text(description='PESEL:')
    email = widgets.Text(description='Email:')
    password = widgets.Password(description='Hasło:')
    button = widgets.Button(description='Dodaj pacjenta')

    def on_click(b):
        try:
            cursor.execute('INSERT INTO patients (name, surname, pesel, email, password) VALUES (?, ?, ?, ?, ?)',
                           (name.value, surname.value, pesel.value, email.value, password.value))
            conn.commit()
            print("✅ Pacjent został dodany.")
        except sqlite3.IntegrityError:
            print("❌ Błąd: pacjent z takim PESEL już istnieje.")

    button.on_click(on_click)
    display(name, surname, pesel, email, password, button)

add_patient_form()
```

```python
def login_form():
    pesel = widgets.Text(description='PESEL:')
    password = widgets.Password(description='Hasło:')
    button = widgets.Button(description='Zaloguj')

    def on_click(b):
        cursor.execute('SELECT * FROM patients WHERE pesel = ? AND password = ?', (pesel.value, password.value))
        result = cursor.fetchone()
        if result:
            print(f"✅ Zalogowano pomyślnie: {result[1]} {result[2]}")
            display_patient_card(result)
        else:
            print("❌ Nieprawidłowe dane logowania.")

    button.on_click(on_click)
    display(pesel, password, button)

login_form()
```

```python
def display_patient_card(patient_data):
    html = f"""
    <h3>Karta pacjenta</h3>
    <ul>
        <li><strong>Imię:</strong> {patient_data[1]}</li>
        <li><strong>Nazwisko:</strong> {patient_data[2]}</li>
        <li><strong>PESEL:</strong> {patient_data[3]}</li>
        <li><strong>Email:</strong> {patient_data[4]}</li>
    </ul>
    """
    display(HTML(html))
```
